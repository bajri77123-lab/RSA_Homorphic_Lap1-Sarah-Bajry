```python
!pip install pycryptodome
# === IoT Sensor Data Encryption Simulation ===
# Part I: Secure Data Transmission for IoT Devices
# Student: [Sarah Bajry]
# Enrollment: [22202041014]
import random
import json
from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes
from Crypto.Util.Padding import pad, unpad
import base64

# --- IoT Device (Sensor) Simulation ---

def generate_sensor_data():
    """Generates random temperature and humidity sensor readings."""
    # Generate random floating-point numbers for sensor readings
    temperature = round(random.uniform(20.0, 40.0), 2)
    humidity = round(random.uniform(30.0, 80.0), 2)

    # Structure the data as a dictionary
    data = {
        "device_id": "IoT-Sensor-001",
        "timestamp": "2025-11-29T10:00:00Z", # Mock timestamp for simplicity
        "temperature_c": temperature,
        "humidity_percent": humidity
    }
    # Convert data dictionary to a JSON string, then encode to bytes
    return json.dumps(data).encode('utf-8')

def encrypt_data(data_bytes, key):
    """Encrypts data using AES in GCM mode (Authenticated Encryption)."""
    # AES GCM is preferred for its integrity check (authentication tag)
    # Create a new cipher object with the key and GCM mode
    cipher = AES.new(key, AES.MODE_GCM)

    # Encrypt the data and generate the authentication tag
    ciphertext, tag = cipher.encrypt_and_digest(data_bytes)

    # The nonce (initialization vector) is needed for decryption
    # The tag is needed for integrity verification
    return cipher.nonce, ciphertext, tag

# --- Server/Base Station Simulation ---

def decrypt_data(nonce, ciphertext, tag, key):
    """Decrypts data using AES in GCM mode and verifies the authentication tag."""
    try:
        # Create a new cipher object with the key, GCM mode, and the received nonce
        cipher = AES.new(key, AES.MODE_GCM, nonce=nonce)

        # Decrypt and verify the tag in one step. If verification fails, a ValueError is raised.
        decrypted_bytes = cipher.decrypt_and_verify(ciphertext, tag)

        # Decode the bytes back to a string
        return decrypted_bytes.decode('utf-8')
    except ValueError:
        # This exception is raised if the tag is invalid (data was tampered with)
        return "Decryption failed: Data integrity check failed (Authentication Tag Invalid)."

# --- Main Simulation Logic ---

def run_part1_simulation():
    """Runs the full IoT data encryption and decryption simulation."""
    print("--- Part I: IoT Device Data Encryption Simulation ---")

    # 1. Generate a secret key (must be 16, 24, or 32 bytes for AES)
    # In a real system, this key would be securely provisioned.
    secret_key = get_random_bytes(16)
    print(f"Secret Key (16 bytes): {base64.b64encode(secret_key).decode('utf-8')}")

    # 2. IoT Device: Generate and prepare data
    raw_data_bytes = generate_sensor_data()
    raw_data_str = raw_data_bytes.decode('utf-8')
    print("\n[IoT Device] Raw Sensor Data (Plaintext):")
    print(raw_data_str)

    # 3. IoT Device: Encrypt the data
    nonce, ciphertext, tag = encrypt_data(raw_data_bytes, secret_key)

    # The device transmits the nonce, ciphertext, and tag to the server
    print("\n[Transmission] Encrypted Data Components (Ciphertext, Nonce, Tag):")
    print(f"Nonce (IV): {base64.b64encode(nonce).decode('utf-8')}")
    print(f"Ciphertext: {base64.b64encode(ciphertext).decode('utf-8')}")
    print(f"Authentication Tag: {base64.b64encode(tag).decode('utf-8')}")

    # 4. Server: Decrypt the data
    decrypted_data_str = decrypt_data(nonce, ciphertext, tag, secret_key)

    print("\n[Server] Decrypted Sensor Data:")
    print(decrypted_data_str)

    # 5. Verification
    if decrypted_data_str == raw_data_str:
        print("\n[Result] SUCCESS: Decrypted data matches the original raw data.")
    else:
        print("\n[Result] FAILURE: Decrypted data does NOT match the original raw data.")

if __name__ == '__main__':
    run_part1_simulation()
```

```python
# --------------------------------------------------------------
# Part III: Secure Boot Verification Simulation (Optional)
# --------------------------------------------------------------

import hashlib
import time

print("=== Secure Boot Verification Simulation ===\n")

stored_hash = "a1b2c3d4e5f67890abcdef1234567890"

firmware_code = """
void setup() {
  Serial.begin(115200);
}
void loop() {
  Serial.println("IoT Device Running...");
}
"""

current_hash = hashlib.sha256(firmware_code.encode()).hexdigest()

print("Stored Firmware Hash : ", stored_hash)
print("Current Firmware Hash:", current_hash)
print("\nVerifying firmware integrity...\n")

time.sleep(1)

if current_hash == stored_hash:
    print("[OK] Secure Boot PASSED ")
else:
    print("[ERROR] Secure Boot FAILED ")
    print("Firmware integrity mismatch — Device Halted.")
```
